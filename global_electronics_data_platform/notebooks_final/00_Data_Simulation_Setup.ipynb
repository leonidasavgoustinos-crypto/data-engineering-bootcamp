```python
import os
import pandas as pd

# --- SIMULATE EXTERNAL SYSTEM DATA EXPORTS ---
# We read the raw JSON files and export them to DBFS in different formats 
# to simulate receiving data from various systems (CRM, ERP, WebShop).

RAW_DATA_PATH = os.path.normpath(os.path.join(os.getcwd(), "..", "data", "raw"))

# 1. Read the original raw JSON files
customers_raw = spark.read.format("json").load(f"file://{RAW_DATA_PATH}/customers.json")
products_raw  = spark.read.format("json").load(f"file://{RAW_DATA_PATH}/products.json")
orders_raw    = spark.read.format("json").load(f"file://{RAW_DATA_PATH}/orders.json")


# 2. Save them into your Workspace Repo in 3 different formats
SIMULATED_PATH = os.path.normpath(os.path.join(os.getcwd(), "..", "data", "simulated"))
os.makedirs(SIMULATED_PATH, exist_ok=True)

# Convert to Pandas AND strip hidden Databricks Connect metadata by recreating the DataFrames
customers_pdf = pd.DataFrame(customers_raw.toPandas().to_dict('list'))
products_pdf  = pd.DataFrame(products_raw.toPandas().to_dict('list'))
orders_pdf    = pd.DataFrame(orders_raw.toPandas().to_dict('list'))

# The CRM System exports data as CSV
customers_pdf.to_csv(os.path.join(SIMULATED_PATH, "customers.csv"), index=False)

# The ERP System exports data as Parquet (optimized)
products_pdf.to_parquet(os.path.join(SIMULATED_PATH, "products.parquet"), index=False)

# The WebShop exports data as JSON (nested structures)
orders_pdf.to_json(os.path.join(SIMULATED_PATH, "orders.json"), orient="records", lines=True)

print("Simulation Complete: CSV, Parquet, and JSON files generated successfully!")
```
