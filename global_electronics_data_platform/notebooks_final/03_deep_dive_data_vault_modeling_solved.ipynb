# 🎓 Educational Deep Dive: Data Vault 2.0

Welcome to the Data Vault Modeling bootcamp! 
Data Vault is an advanced architectural paradigm designed for massive scale and agility. Instead of building rigid Star Schemas, we decouple our entities into three distinct table types:
1. **Hubs:** The core Business Keys (e.g., Customer ID). They never change.
2. **Satellites:** The descriptive context (e.g., Customer Name, City). They change over time (acting like SCD Type 2).
3. **Links:** The relationships connecting Hubs (e.g., an Order linking a Customer and a Product).

Let's learn how to engineer a Data Vault!

## Chapter 1: The Core Architecture Mapping (Theoretical Context)

Before writing data engineering jobs, we map out how operational entities are decoupled into Data Vault primitives to facilitate independent ingestion cycles:

1. **Hub Components (`hub_customer`, `hub_product`):** Capture the immutable Core Business Keys (BKs). They isolate corporate identity independently of source system fluctuations.
2. **Satellite Components (`sat_customer_details`, `sat_product_details`):** Hold the mutable, descriptive contextual attributes (Socio-demographics, pricing states) and store full historicity via system timestamps.
3. **Link Component (`link_order`):** Represents the distinct structural relationships or transactions binding multiple Hub keys together.
4. **Link Satellite Component (`sat_link_order_details`):** Extracted descriptive parameters tightly coupled to the occurrence of a transaction relationship (e.g., non-key transactional dimensions like order dates and line item quantities).

## Chapter 2: Schema Processing Pipeline Execution

### Exercise 2.1: Initializing Staging Data Sources
Read the current operational source data from the internal platform catalog to feed the Data Vault initialization pipeline.

```python
# Read source references from our Bronze layer (Notebook 1)
raw_cust_df = spark.read.table("bronze_customers")
raw_orders_df = spark.read.table("bronze_orders")
raw_prod_df = spark.read.table("bronze_products")
```

### Exercise 2.2: Deploying Business Hubs (Identity Tables)
Generate deterministic SHA-256 hash keys to create separate identity structures for Customer and Product business entities.

```python
%sql
USE CATALOG ai_lab;
USE SCHEMA default;
```

```python
from pyspark.sql import functions as F

# 1. CUSTOMER HUB
# Hubs store ONLY the Business Key and its Hash
hub_customers_df = raw_cust_df.withColumn(
    "customer_hash_key", F.sha2(F.upper(F.trim(F.col("customer_id").cast("string"))), 256)
).select(
    "customer_hash_key", 
    "customer_id", 
    F.current_timestamp().alias("load_datetime"), 
    F.lit("CRM").alias("record_source")
).distinct()

hub_customers_df.write.format("delta").mode("overwrite").saveAsTable("hub_customer")
display(spark.read.table("hub_customer"))
```

### 👨‍🏫 Instructor Note: Hubs & Hashing
In Data Vault 2.0, we rely heavily on Cryptographic Hashing to generate Primary Keys (Surrogate Keys). 
**Why?** Because hashing is deterministic. If two source systems send the same Business Key independently, they generate the exact same Hash Key. This allows us to load tables in parallel without sequence generators!
**Solution:** We use `F.sha2(..., 256)` to hash the `product_id`.

```python
# 2. PRODUCT HUB 
hub_products_df = raw_prod_df.withColumn(
    "product_hash_key", F.sha2(F.upper(F.trim(F.col("product_id").cast("string"))), 256)
).select(
    "product_hash_key", 
    "product_id", 
    F.current_timestamp().alias("load_datetime"), 
    F.lit("ERP").alias("record_source")
).distinct()

hub_products_df.write.format("delta").mode("overwrite").saveAsTable("hub_product")
display(spark.read.table("hub_product"))
```

### Exercise 2.3: Deploying Descriptive Context Satellites
Build the descriptive context table structures. Satellites map changes over time back to their respective parent Hash Key identifiers.

```python
# 3. PRODUCT SATELLITE 
# Satellites store the descriptive attributes and link back to the Hub via the hash key
sat_products_df = raw_prod_df.withColumn(
    "product_hash_key", F.sha2(F.upper(F.trim(F.col("product_id").cast("string"))), 256)
).select(
    "product_hash_key", # Links to Hub Product
    "product_name", 
    "category", 
    "price", 
    F.current_timestamp().alias("load_datetime"), 
    F.lit("ERP").alias("record_source")
)

sat_products_df.write.format("delta").mode("overwrite").saveAsTable("sat_product_details")
display(spark.read.table("sat_product_details"))
```

```python
# 4. CUSTOMER SATELLITE 
sat_customer_df = raw_cust_df.withColumn(
    "customer_hash_key", F.sha2(F.upper(F.trim(F.col("customer_id").cast("string"))), 256)
).select(
    "customer_hash_key", # Links to Hub Customer
    "customer_name", 
    "city", 
    F.current_timestamp().alias("load_datetime"), 
    F.lit("CRM").alias("record_source")
)

sat_customer_df.write.format("delta").mode("overwrite").saveAsTable("sat_customer_details")
display(spark.read.table("sat_customer_details"))
```

### Exercise 2.4: Deploying Structural Transaction Links
Flatten multi-entity relationships by building a clean connection entity bridging the Customer Hub and Product Hub targets.

### 👨‍🏫 Instructor Note: Links & Concatenation
A Link represents a transaction or relationship between multiple Hubs. To create the unique Hash Key for the Link itself, we must hash the combination of all participating Business Keys.
**Solution:** We concatenate the keys using a standard double-pipe delimiter `||`. This prevents 'Hash Collisions' (e.g., preventing "12" and "3" from generating the same hash as "1" and "23").

```python
# To successfully map relationships, we flatten nested transactional items first
raw_orders_exploded = raw_orders_df.withColumn("item", F.explode("items")) \
    .withColumn("product_id", F.col("item.product_id")) \
    .withColumn("quantity", F.col("item.quantity"))

# 5. ORDER LINK 
# Links connect Hubs together. We concatenate the IDs with "||" and hash the result.
link_orders_df = raw_orders_exploded.withColumn(
    "link_order_hash_key",
    F.sha2(F.concat_ws("||", 
        F.upper(F.trim(F.coalesce(F.col("order_id").cast("string"), F.lit("")))),
        F.upper(F.trim(F.coalesce(F.col("customer_id").cast("string"), F.lit("")))),
        F.upper(F.trim(F.coalesce(F.col("product_id").cast("string"), F.lit(""))))
    ), 256)
).withColumn(
    "customer_hash_key", F.sha2(F.upper(F.trim(F.col("customer_id").cast("string"))), 256)
).withColumn(
    "product_hash_key", F.sha2(F.upper(F.trim(F.col("product_id").cast("string"))), 256)
).select(
    "link_order_hash_key", 
    "order_id", 
    "customer_hash_key", 
    "product_hash_key", 
    F.current_timestamp().alias("load_datetime"), 
    F.lit("WEB_SHOP").alias("record_source")
).distinct()

link_orders_df.write.format("delta").mode("overwrite").saveAsTable("link_order")
display(spark.read.table("link_order"))
```

### Exercise 2.5: Deploying a Satellite on top of a Link
Because transactional relationships hold descriptive indicators (order dates, volume units), you must attach an operational Satellite directly to the transaction Link.

```python
# 6. SATELLITE ON LINK
# Stores the descriptive attributes of the relationship (e.g. order date, quantity)
sat_link_orders_df = raw_orders_exploded.withColumn(
    "link_order_hash_key",
    F.sha2(F.concat_ws("||", 
        F.upper(F.trim(F.coalesce(F.col("order_id").cast("string"), F.lit("")))),
        F.upper(F.trim(F.coalesce(F.col("customer_id").cast("string"), F.lit("")))),
        F.upper(F.trim(F.coalesce(F.col("product_id").cast("string"), F.lit(""))))
    ), 256)
).select(
    "link_order_hash_key", # Links to Link Order
    "order_date",
    "quantity",
    F.current_timestamp().alias("load_datetime"),
    F.lit("WEB_SHOP").alias("record_source")
)

sat_link_orders_df.write.format("delta").mode("overwrite").saveAsTable("sat_link_order_details")
display(spark.read.table("sat_link_order_details"))

print("Data Vault Architecture Status: Enterprise Layout Fully Sampled and Generated.")
```

### Data Vault SCD Type 2 (Insert-Only Architecture)
In the Data Vault methodology, Satellites handle history tracking by **never updating or deleting** old rows. Instead of comparing columns one by one, we generate a `hash_diff` (a cryptographic signature of all descriptive columns). 
If the incoming `hash_diff` doesn't match the latest one in our Satellite, it means a change occurred. We simply **INSERT** the new row with a fresh `load_datetime`. The active record is always the one with the maximum `load_datetime`.

```python
from pyspark.sql import functions as F
from delta.tables import DeltaTable
from pyspark.sql.types import StructType, StructField, LongType, StringType

# 1. Day 2 Data Preparation
target_df = spark.read.table("silver_dim_customers")

# Dynamically get the very first real customer ID and Name to ensure an update happens
real_cust_id = target_df.select("customer_id").first()[0]
real_cust_name = target_df.select("customer_name").first()[0]

# Create the new batch: The old customer changed city, and we added a brand new one
incoming_df = spark.createDataFrame([
    (real_cust_id, real_cust_name, "Patras (MOVED!)"), # This will trigger an UPDATE (Insert new state)
    (999999, "Eleni", "Thessaloniki")                  # This will trigger an INSERT
], ["customer_id", "customer_name", "city"])

print("Incoming Delta Staging Batch (Day 2):")
display(incoming_df)
```

```python
# 2. Generate Hash Keys and Hash Diff for incoming data
incoming_sat_df = incoming_df.withColumn(
    "customer_hash_key", F.sha2(F.upper(F.trim(F.col("customer_id").cast("string"))), 256)
).withColumn(
    # The Hash Diff combines all descriptive columns. If city changes, the hash completely changes!
    "hash_diff", F.sha2(F.concat_ws("||", 
        F.upper(F.trim(F.col("customer_name"))), 
        F.upper(F.trim(F.col("city")))
    ), 256)
).select(
    "customer_hash_key",
    "hash_diff",
    "customer_name", 
    "city",
    F.current_timestamp().alias("load_datetime"), 
    F.lit("CRM").alias("record_source")
)

# Dynamically alter the target table to support the hash_diff column if it doesn't exist
try:
    spark.sql("ALTER TABLE sat_customer_details ADD COLUMNS (hash_diff STRING)")
    # Backfill hash_diff for existing records
    spark.sql("""
        UPDATE sat_customer_details 
        SET hash_diff = SHA2(CONCAT_WS('||', UPPER(TRIM(customer_name)), UPPER(TRIM(city))), 256)
    """)
except Exception as e:
    pass # Column already exists, proceed normally


# 3. Execute Data Vault SCD Type 2 (Insert Only)
print("Data Vault SCD Type 2 - Insert Only Execution:")

(
    DeltaTable.forName(spark, "sat_customer_details").alias("target")
    # We match on both the Hub Key AND the Hash Diff
    .merge(incoming_sat_df.alias("source"), 
           "target.customer_hash_key = source.customer_hash_key AND target.hash_diff = source.hash_diff")
    # If the hash_diff is NOT found, it means the customer is either brand new or changed city!
    # Therefore, we simply INSERT the new state. (No UPDATE is performed).
    .whenNotMatchedInsert(values = {
        "customer_hash_key": "source.customer_hash_key",
        "hash_diff": "source.hash_diff",
        "customer_name": "source.customer_name",
        "city": "source.city",
        "load_datetime": "source.load_datetime",
        "record_source": "source.record_source"
    })
    .execute()
)

# Display the satellite to show multiple history rows for the same customer
display(spark.read.table("sat_customer_details").orderBy("customer_name", "load_datetime"))
```
